```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("tested.csv")
```

```python
print("first 5 Rows:")
print(df.head())
print("Last 5 Rows:")
print(df.tail())
print("Dataset shape")
print(df.shape)
print("Dataset informatio ")
print(df.info())
print("Statistical escription ")
print(df.describe())
```

```python

print("Column Names:")
print(df.columns)
print("Duplicate Records ")
print(df.duplicated().sum())
print("survival Count ")
print(df["Survived"].value_counts())

```

```python
print("Missing values before cleaning ")
print(df.isnull().sum())
```

```python
df = df.drop(columns=["Cabin"], errors="ignore")
```

```python
df["Age"] = df["Age"].fillna(
    df["Age"].median()
)
```

```python
df["Embarked"] = df["Embarked"].fillna(
    df["Embarked"].mode()[0]
)
```

```python
df["Fare"] = df["Fare"].fillna(
    df["Fare"].median()
)
```

```python
print("missing values after Cleaning ")
print(df.isnull().sum())
print("\nt otal missing values ")
print(df.isnull().sum().sum())
```

```python
print("average age ")
print(df["Age"].mean())
print("Median age ")
print(df["Age"].median())

print("average fare",df["Fare"].mean())
print("survival rat", df["Survived"].mean() * 100)


```

```python
sns.countplot(data=df, x="Survived")
plt.title("passenger survival count")
plt.show()
```

```python
sns.countplot(data=df, x="Sex", hue="Survived")
plt.title("survival based on Gender")
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.histplot(df["Age"], bins=30, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
sns.countplot(data=df, x="Embarked", hue="Survived")

plt.title("Survival Based on Embarkation Location")
plt.show()
```

```python
print("Final Dataset:")
print(df.head())
print("Final Dataset Shape:", df.shape)
df.to_csv("titanic_cleaned.csv", index=False)

print("Cleaned Dataset Saved Successfully!")
```
